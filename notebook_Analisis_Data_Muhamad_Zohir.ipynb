## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

**Pertanyaan Bisnis**
- **Pertanyaan 1:** Bagaimana tren bulanan jumlah pesanan dan total nilai penjualan pada periode Januari 2017 hingga Agustus 2018, serta bulan apa yang mencatat nilai penjualan tertinggi dan terendah?
- **Pertanyaan 2:** Bagaimana rata-rata durasi pengiriman pesanan dan persentase pesanan yang terlambat dibandingkan dengan tanggal estimasi pengiriman pada periode Januari hingga Agustus 2018?
- **Pertanyaan 3:** Bagaimana distribusi pelanggan berdasarkan segmen Recency, Frequency, dan Monetary (RFM) selama periode transaksi yang tersedia?

### Assessing Data

#### Identifying ... problem

```python
#Membuat Fungsi Assessing Data
def checking_data(dataset, nama_dataset, kolom):
      print(f"=== Missing Value {nama_dataset}===")
      print(dataset.isna().sum())
      print("=" * 60)

      print(f"=== Duplicated Value {nama_dataset} ===")
      print(dataset.duplicated().sum())
      print("=" * 60)

      for col in kolom:
        print(f"=== Unique Value {nama_dataset} - {col} ===")
        print(dataset[col].unique())
        print("=" * 60)
        print()

def checking_outliers(dataset, name_dataset, kolom):
  print(f"=== Outliers Dataset {name_dataset} - {kolom} ===")
  Q1 = dataset[kolom].quantile(0.25)
  Q3 = dataset[kolom].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = dataset[(dataset[kolom] < lower_bound) | (dataset[kolom] > upper_bound)]
  print(f"Jumlah Outliers: {len(outliers)}")
  print("="*60)
  print("\n")
  return outliers, lower_bound, upper_bound
```

```python
#INTEGRASI DATA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Konversi kolom tanggal menjadi datetime
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_dataset[col] = pd.to_datetime(
        orders_dataset[col],
        errors="coerce"
    )

# Menghitung total harga produk setiap pesanan
order_sales = (
    order_items_dataset
    .groupby("order_id", as_index=False)
    .agg(total_sales=("price", "sum"))
)

# Integrasi pesanan dengan total harga produk
df_analysis = orders_dataset.merge(
    order_sales,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Integrasi identitas pelanggan
df_analysis = df_analysis.merge(
    customers_dataset[
        ["customer_id", "customer_unique_id"]
    ],
    on="customer_id",
    how="left",
    validate="many_to_one"
)

print("Jumlah baris orders:", len(orders_dataset))
print("Jumlah baris hasil integrasi:", len(df_analysis))
print("Jumlah order unik:", df_analysis["order_id"].nunique())
print(
    "Jumlah customer unik:",
    df_analysis["customer_unique_id"].nunique()
)

df_analysis.head()
```

```python
#Validasi hasil integrasi
print(
    "Jumlah order unik sebelum integrasi:",
    orders_dataset["order_id"].nunique()
)

print(
    "Jumlah order unik setelah integrasi:",
    df_analysis["order_id"].nunique()
)

print("\nMissing value setelah integrasi:")
print(
    df_analysis[
        ["total_sales", "customer_unique_id"]
    ].isna().sum()
)

print("\nDistribusi status pesanan:")
print(
    df_analysis["order_status"].value_counts(
        dropna=False
    )
)
```

```python
# Mengambil order_id yang tidak memiliki total_sales
missing_sales = df_analysis[
    df_analysis["total_sales"].isna()
].copy()

print("Jumlah order tanpa total_sales:", len(missing_sales))

# Memeriksa apakah order_id tersebut tersedia di order_items_dataset
item_order_ids = order_items_dataset["order_id"].unique()

missing_sales["has_order_items"] = (
    missing_sales["order_id"].isin(item_order_ids)
)

print("\nKetersediaan item pada order:")
print(missing_sales["has_order_items"].value_counts())

print("\nDistribusi status pesanan:")
print(missing_sales["order_status"].value_counts())

missing_sales[
    ["order_id", "order_status", "has_order_items"]
].head(10)
```

```python
# Memeriksa apakah price kosong pada seluruh tabel item
print(
    "Missing price pada order_items_dataset:",
    order_items_dataset["price"].isna().sum()
)

# Memeriksa apakah ada order_id duplikat dalam hasil agregasi
print(
    "Duplikat order_id pada order_sales:",
    order_sales["order_id"].duplicated().sum()
)

# Memeriksa sampel item untuk order yang memiliki masalah
problem_order_ids = missing_sales.loc[
    missing_sales["has_order_items"],
    "order_id"
]

print(
    order_items_dataset[
        order_items_dataset["order_id"].isin(problem_order_ids)
    ].head(10)
)
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
#Menyiapkan data analisis penjualan
df_q1 = df_analysis[
    df_analysis["order_status"].eq("delivered")
    & df_analysis["order_purchase_timestamp"].ge(
        pd.Timestamp("2017-01-01")
    )
    & df_analysis["order_purchase_timestamp"].lt(
        pd.Timestamp("2018-09-01")
    )
].copy()

print("Jumlah order untuk analisis:", df_q1["order_id"].nunique())
print("Jumlah data dengan total_sales kosong:", df_q1["total_sales"].isna().sum())
```

```python
#Menghitung jumlah order dan total penjualan per bulan
df_q1["month"] = (
    df_q1["order_purchase_timestamp"].dt.to_period("M")
)

monthly_sales = (
    df_q1.groupby("month")
    .agg(
        total_orders=("order_id", "nunique"),
        total_sales=("total_sales", "sum")
    )
    .reindex(
        pd.period_range("2017-01", "2018-08", freq="M"),
        fill_value=0
    )
)

monthly_sales.index.name = "month"

print(monthly_sales)
```

```python
#Visualisasi tren bulanan
fig, ax1 = plt.subplots(figsize=(14, 6))

ax1.plot(
    monthly_sales.index.astype(str),
    monthly_sales["total_orders"],
    marker="o",
    label="Jumlah Order"
)

ax1.set_xlabel("Bulan")
ax1.set_ylabel("Jumlah Order")
ax1.tick_params(axis="x", rotation=60)

ax2 = ax1.twinx()

ax2.plot(
    monthly_sales.index.astype(str),
    monthly_sales["total_sales"],
    marker="s",
    linestyle="--",
    label="Total Penjualan"
)

ax2.set_ylabel("Total Penjualan")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc="upper left"
)

plt.title("Tren Bulanan Jumlah Order dan Total Penjualan")
plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

```python
#Menyiapkan data customer
df_rfm_source = df_analysis[
    df_analysis["order_status"].eq("delivered")
].copy()

df_rfm_source = df_rfm_source.dropna(
    subset=[
        "customer_unique_id",
        "order_purchase_timestamp",
        "total_sales"
    ]
)

print("Jumlah order untuk RFM:", len(df_rfm_source))
print(
    "Jumlah customer unik untuk RFM:",
    df_rfm_source["customer_unique_id"].nunique()
)
```

```python
#Menghitung nilai Recency, Frequency, dan Monetary
##Tanggal acuan adalah satu hari setelah transaksi terakhir
reference_date = (
    df_rfm_source["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

rfm = df_rfm_source.groupby("customer_unique_id").agg(
    last_purchase=("order_purchase_timestamp", "max"),
    Frequency=("order_id", "nunique"),
    Monetary=("total_sales", "sum")
)

rfm["Recency"] = (
    reference_date - rfm["last_purchase"]
).dt.days

rfm = rfm[["Recency", "Frequency", "Monetary"]]

print(rfm.head())
print("\nStatistik deskriptif RFM:")
print(rfm.describe())
```

```python
#Memberikan skor RFM
def quartile_score(series):
    ranks = series.rank(method="first")

    return pd.qcut(
        ranks,
        q=4,
        labels=[1, 2, 3, 4]
    ).astype(int)


rfm["R_score"] = 5 - quartile_score(rfm["Recency"])
rfm["F_score"] = quartile_score(rfm["Frequency"])
rfm["M_score"] = quartile_score(rfm["Monetary"])

rfm["RFM_score"] = (
    rfm["R_score"].astype(str)
    + rfm["F_score"].astype(str)
    + rfm["M_score"].astype(str)
)

print(rfm.head(10))
```

```python
#Membuat segmen customer
def assign_segment(row):
    r = row["R_score"]
    f = row["F_score"]
    m = row["M_score"]

    if r >= 3 and f >= 3 and m >= 3:
        return "Champions"
    elif f >= 3 and r >= 2:
        return "Loyal Customers"
    elif r >= 3 and f <= 2:
        return "Recent Customers"
    elif r <= 2 and f >= 2:
        return "At Risk"
    elif m >= 3:
        return "Big Spenders"
    else:
        return "Others"


rfm["Segment"] = rfm.apply(assign_segment, axis=1)

print(rfm.head(10))
```

```python
#Ringkasan jumlah dan persentase customer per segmen
segment_summary = (
    rfm.groupby("Segment")
    .agg(
        total_customers=("Segment", "size"),
        avg_recency=("Recency", "mean"),
        avg_frequency=("Frequency", "mean"),
        avg_monetary=("Monetary", "mean")
    )
    .sort_values("total_customers", ascending=False)
)

segment_summary["customer_percentage"] = (
    segment_summary["total_customers"]
    / segment_summary["total_customers"].sum()
    * 100
)

print(segment_summary.round(2))
```

```python
#Visualisasi distribusi segmen customer
segment_counts = rfm["Segment"].value_counts()

plt.figure(figsize=(10, 6))

plt.bar(
    segment_counts.index,
    segment_counts.values
)

plt.title("Distribusi Customer Berdasarkan Segmen RFM")
plt.xlabel("Segmen Customer")
plt.ylabel("Jumlah Customer")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
